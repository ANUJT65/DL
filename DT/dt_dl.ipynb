```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

# Load data
iris = load_iris()
X, y = iris.data, iris.target

# Define and train the classifier, using 'clas' as the variable name
clas = DecisionTreeClassifier(random_state=42)
clas.fit(X, y)

# Now you can make predictions with the classifier
observation = [[5, 4, 3, 2]]  # Example observation
predicted_class = clas.predict(observation)
predicted_proba = clas.predict_proba(observation)

print("Predicted Class:", predicted_class)
print("Prediction Probabilities:", predicted_proba)
```

```python
plot_tree(clas)
```

```python
viz = dtreeviz.model(clas, X_train, y_train, feature_names=iris.feature_names,
               class_names=["setosa", "versicolor", "virginica"])
viz
```
